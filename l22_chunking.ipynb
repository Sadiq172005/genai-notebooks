```python
from langchain_pdfmuse import PdfmuseLoader

# one Document per page (default)
docs = PdfmuseLoader("data/sampled_schemes.pdf").load()
```

```python
len(docs)
```

```python
print(docs[0].page_content)  # first 100 characters of the first page
```

```python
docs
```

```python
from langchain_text_splitters import CharacterTextSplitter

text_splitter = CharacterTextSplitter.from_tiktoken_encoder(
    encoding_name="cl100k_base", chunk_size=200, chunk_overlap=0
)
texts = text_splitter.split_documents(docs)
```

```python
len(texts)
```

```python
texts
```

```python
print(texts[7].page_content)  # first 100 characters of the first chunk
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=0)
c_texts = text_splitter.split_documents(docs)
```
