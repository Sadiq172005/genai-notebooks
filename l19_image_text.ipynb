# Image Input with LangChain

LangChain represents image input as a `{"type": "image", ...}` **content block** inside a message's `content` list, alongside `{"type": "text", ...}` blocks. There are two ways to supply the image:

- `base64` — inline the raw image bytes as a base64-encoded string, plus a `mime_type`.
- `url` — give a publicly reachable URL; the provider fetches it directly.

This notebook shows both.

```python
import base64
from pathlib import Path

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage

load_dotenv()

llm = ChatOpenAI(model="gpt-5.6-luna")
```

```python
from pathlib import Path

path = Path("pc1.png")
```

```python
image_bytes = path.read_bytes()  # Now opens and reads the file
base64_bytes = base64.b64encode(image_bytes)
```

```python
def encode_image(path):
    # Create a Path object
    image_path = Path(path)

    # Read the image as bytes
    image_bytes = image_path.read_bytes()

    # Encode the bytes into Base64 (still bytes)
    base64_bytes = base64.b64encode(image_bytes)

    # Convert Base64 bytes into a UTF-8 string
    base64_string = base64_bytes.decode("utf-8")

    return base64_string
```

```python
message =[ HumanMessage(
    content=[
        {"type": "text", "text": "tell me the name of the person in that icard"},
        {
            "type": "image",
            "base64": encode_image("pc1.png"),
            "mime_type": "image/png",
        },
    ]
)]

response = llm.invoke(message)
print(response.content)
```

```python
message =[ HumanMessage(
    content=[
        {"type": "text", "text": "tell me the school name "},
        {
            "type": "image",
            "base64": encode_image("id1.png"),
            "mime_type": "image/png",
        },
    ]
)]

response = llm.invoke(message)
print(response.content)
```

**Why base64 is required here:** `p1.png` only exists as bytes on your local disk — the OpenAI API has no way to reach into your filesystem. The only way to get those bytes to the model is to encode them as base64 text and embed them directly in the JSON request body sent over HTTPS.

```python
message = HumanMessage(
    content=[
        {"type": "text", "text": "What is happening in this picture? Who or what is the main subject?"},
        {
            "type": "image",
            "base64": encode_image("pc2.png"),
            "mime_type": "image/png",
        },
    ]
)

response = llm.invoke([message])
print(response.content)
```

## Passing an image by URL

If the image is already hosted somewhere publicly reachable, you can skip encoding entirely and just pass its `url`.

```python
message = HumanMessage(
    content=[
        {"type": "text", "text": "which animal is there and what is its color"},
        {
            "type": "image",
            "url": "https://upload.wikimedia.org/wikipedia/commons/3/3a/Cat03.jpg",
        },
    ]
)

response = llm.invoke([message])
print(response.content)
```
