* A Pydantic Schema is simply a Python class that describes the structure of your data.

* Think of it as a template or blueprint.

```python
from pydantic import BaseModel, Field
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import SystemMessage, HumanMessage, AIMessage

load_dotenv()

llm = ChatOpenAI(model="gpt-5.4-nano")
```

```python
from pydantic import BaseModel, Field

class Db_selector(BaseModel):
    database_name:list[str]
    database_description:str
```

```python
description_context=""""
1. MovieDB
   Description: Stores movies, actors, directors, genres, release dates, ratings, reviews, and box office collections.

2. LibraryDB
   Description: Manages books, authors, publishers, library members, borrowing records, and return dates.

3. HospitalDB
   Description: Contains patient records, doctors, appointments, diagnoses, prescriptions, and medical history.

4. ECommerceDB
   Description: Stores customer profiles, products, orders, payments, inventory, and shipping details.

5. UniversityDB
   Description: Manages students, professors, departments, courses, enrollments, and examination results.

6. BankingDB
   Description: Stores customer accounts, transactions, loans, credit cards, and account balances.

7. HotelManagementDB
   Description: Contains hotel rooms, guests, reservations, staff information, and billing records.

8. EmployeeManagementDB
   Description: Manages employee details, departments, salaries, attendance, leave records, and performance reviews.

9. TravelBookingDB
   Description: Stores flights, hotels, travel packages, passengers, bookings, and payment information.

10. MusicStreamingDB
    Description: Contains songs, artists, albums, playlists, users, subscriptions, and listening history.
"""
```

```python

template1 = [
    SystemMessage(f"choose the right datbase from this {description_context} for user query"
    ),
    HumanMessage("iwant to know about,flipkart and sbi bank")]
```

```python
llm_with_schema= llm.with_structured_output(Db_selector)
```

```python
output = llm_with_schema.invoke(template1)
```

```python
output.database_name[1]
```

```python
from pydantic import BaseModel, Field

class Candidate(BaseModel):
    """Candidate information extracted from a resume."""

    name: str = Field(description="Candidate's full name")
    email: str = Field(description="Candidate's email address")
    phone: str = Field(description="Candidate's phone number")
    experience: int = Field(description="Years of experience")
    skills: list[str] = Field(description="Technical skills")
```

```python
candidate_llm = llm.with_structured_output(Candidate)


input_msg = [
    HumanMessage("""
Rahul Sharma has worked as a Python Developer for 6 years.
His skills include Python, FastAPI, AWS, Docker, and LangChain.
Email: rahul@gmail.com
Phone: +91-9876543210
"""),
]

result = candidate_llm.invoke(input_msg)

print(result)
```

```python
from pydantic import BaseModel, Field

class Invoice(BaseModel):
    invoice_number: str
    vendor: str
    invoice_date: str
    total_amount: float
    gst_number: str
```

```python
invoice_llm = llm.with_structured_output(Invoice)

result = invoice_llm.invoke("""
Invoice No: INV-2026-001
Vendor: ABC Technologies Pvt Ltd
GSTIN: 29ABCDE1234F1Z5
Date: 25 July 2026
Total Amount: ₹58,450
""")

print(result)
```

```python
from pydantic import BaseModel, Field


class MovieDetails(BaseModel):
    """Details about a movie."""

    title: str = Field(description="The title of the movie")
    year: int = Field(description="The year the movie was released")
    director: str = Field(description="The director of the movie")
    rating: float = Field(description="The movie's rating out of 10")


movie_llm = llm.with_structured_output(MovieDetails)
result = movie_llm.invoke("Tell me about the movie Inception")

print(type(result))
result
```

```python
class Actor(BaseModel):
    name: str = Field(description="The actor's name")
    role: str = Field(description="The character they played (villian, hero or just side actor")

class MovieCast(BaseModel):
    """A movie with its main cast."""

    title: str = Field(description="The title of the movie")
    year: int = Field(description="The year the movie was released")
    cast: list[Actor] = Field(description="The main cast of the movie")


cast_llm = llm.with_structured_output(MovieCast)
cast_result = cast_llm.invoke("Who starred in The Spiderman 3 Movie")
cast_result
```

```python
from typing_extensions import Annotated, TypedDict


class MovieDetailsDict(TypedDict):
    """Details about a movie."""

    title: Annotated[str, ..., "The title of the movie"]
    year: Annotated[int, ..., "The year the movie was released"]
    director: Annotated[str, ..., "The director of the movie"]
    rating: Annotated[float, ..., "The movie's rating out of 10"]


movie_dict_llm = llm.with_structured_output(MovieDetailsDict)
dict_result = movie_dict_llm.invoke("Tell me about the movie Spider-Man: Brand New Universe")

print(type(dict_result))
dict_result
```

```python
from langchain_core.messages import SystemMessage, HumanMessage

template1 = [
    SystemMessage(
        """
You are a movie information extractor.
Return ONLY valid JSON using this schema:

{
  "title": "string",
  "year": 0,
  "genre": ["string"],
  "director": "string",
  "cast": [{"name": "string", "role": "string"}],
  "summary": "string",
  "rating": 0.0
}
"""
    ),
    HumanMessage(
        "The Dark Knight is a 2008 superhero film directed by Christopher Nolan..."
    ),
]

response = llm.invoke(template1)
print(response.content)
```
