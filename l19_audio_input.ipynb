```python
import base64
from pathlib import Path

from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain.messages import HumanMessage

load_dotenv()

llm = ChatOpenAI(model="gpt-audio-mini")
```

```python
def encode_file(path: str) -> str:
    return base64.b64encode(Path(path).read_bytes()).decode("utf-8")
```

```python
message = HumanMessage(
    content=[
        {"type": "text", "text": "whos the speaker in this speech give me biography of the speaker and the speech"},
        {
            "type": "audio",
            "base64": encode_file("clip.mp3"),
            "mime_type": "audio/mp3",
        },
    ]
)

response = llm.invoke([message])
print(response.content)
```

```python
import requests

audio_bytes = requests.get(
    "https://openaiassets.blob.core.windows.net/$web/API/docs/audio/alloy.wav"
).content
audio_base64 = base64.b64encode(audio_bytes).decode("utf-8")

message = HumanMessage(
    content=[
        {"type": "text", "text": "What is being said in this audio clip?"},
        {
            "type": "audio",
            "base64": audio_base64,
            "mime_type": "audio/wav",
        },
    ]
)

response = llm.invoke([message])
print(response.content)
```
